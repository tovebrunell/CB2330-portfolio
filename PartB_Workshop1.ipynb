```python
def reg_GC(seq):
  n = 0
  for nuc in seq:
    if nuc == "G" or nuc == "C":
      n = n + 1

  fraction = n/len(seq)

  return (fraction)
```

```python
import numpy as np
from Bio.SeqUtils import gc_fraction

def num_GC(seq):
  fraction = gc_fraction(seq)
  return fraction
```
